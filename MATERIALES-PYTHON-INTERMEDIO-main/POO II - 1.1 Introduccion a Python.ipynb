## 1. ¿Qué es Python? (10 min)

Python es un lenguaje de programación de **alto nivel**, **interpretado** y de **propósito general**. Lo creó Guido van Rossum y su primera versión salió en 1991. Su diseño prioriza que el código sea **fácil de leer**.

**¿Dónde se usa?** Desarrollo web (Django, Flask), ciencia de datos e inteligencia artificial (pandas, scikit-learn), automatización, aplicaciones de escritorio (**PyQt**, que usaremos en la Unidad 2) y pruebas de software.

### Python frente a Java (lo que ya saben)

| Característica | Java | Python |
|---|---|---|
| Ejecución | Se compila (`javac`) a bytecode y se ejecuta en la JVM | El intérprete ejecuta el código directamente |
| Tipos | **Estáticos**: `int edad = 20;` | **Dinámicos**: `edad = 20` |
| Bloques | Llaves `{ }` | **Indentación** (4 espacios) |
| Fin de sentencia | Punto y coma `;` | Salto de línea |
| Punto de entrada | `public static void main(String[] args)` | El archivo se ejecuta de arriba hacia abajo |
| Imprimir | `System.out.println("Hola");` | `print("Hola")` |
| Comentarios | `//` y `/* */` | `#` y cadenas `""" """` |
| Convención de nombres | `camelCase` | `snake_case` |

> 💡 **Idea clave:** la lógica que aprendieron en Java (variables, condiciones, ciclos, objetos) **sigue siendo la misma**. Lo que cambia es la **sintaxis**.

## 4. Variables y tipado dinámico (15 min)

En Java se **declara el tipo** de cada variable. En Python **no**: el tipo lo determina el valor asignado.

| Java | Python |
|---|---|
| `String nombre = "Ana";` | `nombre = "Ana"` |
| `int edad = 20;` | `edad = 20` |
| `double saldo = 150.75;` | `saldo = 150.75` |
| `boolean activo = true;` | `activo = True` |

⚠️ En Python, `True` y `False` se escriben **con mayúscula inicial**.

````python
nombre = "Ana Pérez"
edad = 20
saldo = 150.75
activo = True

print(nombre, edad, saldo, activo)
````

La función `type()` indica el tipo de dato de una variable:

````python
print(type(nombre))
print(type(edad))
print(type(saldo))
print(type(activo))
````

### Tipado dinámico
Una misma variable puede cambiar de tipo. Esto **no** se puede hacer en Java:

````python
dato = 10
print(dato, type(dato))

dato = "diez"
print(dato, type(dato))

dato =True
print(dato, type(dato))
````

> ✅ **Buena práctica:** aunque Python lo permite, evite cambiar el tipo de una variable, porque el código se vuelve confuso.

### Reglas para nombrar variables
- Pueden contener letras, números y `_`, pero **no pueden empezar con un número**.
- Distinguen mayúsculas y minúsculas (`saldo` ≠ `Saldo`).
- No pueden ser palabras reservadas (`if`, `for`, `class`, `def`...).
- Convención en Python: **snake_case** → `saldo_inicial`, `numero_cuenta` (en Java era `saldoInicial`).

### Asignación múltiple

````python
cuenta, tipo, tasa = "AH-001", "Ahorros", 0.05
print(cuenta, tipo, tasa)

# Intercambiar valores sin variable auxiliar (en Java necesitábamos un temp)
a, b = 1, 2
a, b = b, a
print("a =", a, "b =", b)
````

### f-strings: la forma moderna de armar mensajes
En Java concatenábamos con `+`. En Python, una **f-string** (texto precedido de `f`) permite insertar variables entre llaves `{ }`:

````python
socio = "Luis Torres"
saldo = 1234.5

print("Socio: " + socio + " | Saldo: " + str(saldo))   # concatenando (hay que convertir a str)
print(f"Socio: {socio} | Saldo: {saldo}")               # f-string
print(f"Saldo con 2 decimales: ${saldo:.2f}")           # formato numérico
````

⚠️ **Error frecuente:** concatenar texto con un número sin convertirlo produce un `TypeError`. Ejecute la celda y lea el mensaje de error:

````python
print("Saldo: " + str(saldo))  # This line has a typo: 'srt' should be 'str'
````

## 6. Operadores (10 min)

### Aritméticos
| Operación | Java | Python | Ejemplo Python | Resultado |
|---|---|---|---|---|
| Suma / resta / multiplicación | `+ - *` | `+ - *` | `7 * 2` | `14` |
| División | `7 / 2` → `3` (si ambos son int) | `/` **siempre decimal** | `7 / 2` | `3.5` |
| División entera | `7 / 2` con int | `//` | `7 // 2` | `3` |
| Residuo | `%` | `%` | `7 % 2` | `1` |
| Potencia | `Math.pow(2, 3)` | `**` | `2 ** 3` | `8` |

````python
print(7 / 2)    # división real
print(7 // 2)   # división entera
print(7 % 2)    # residuo
print(2 ** 3)   # potencia
````

### Relacionales y lógicos
Los relacionales son iguales a Java (`==`, `!=`, `<`, `>`, `<=`, `>=`). Los lógicos **se escriben con palabras**:

| Java | Python |
|---|---|
| `&&` | `and` |
| `\|\|` | `or` |
| `!` | `not` |

Python no tiene `i++` ni `i--`; se usa `i += 1`.

````python
edad = int(input("Ingrese su edad: "))
saldo = int(input("Ingrese su saldo: "))

print(edad >= 18 and saldo >= 100)   # False: el saldo no alcanza
print(edad >= 18 or saldo >= 100)    # True
print(not (edad >= 18))              # False

contador = 0
contador += 1     # en Java: contador++
print(contador)
````

Python también permite **encadenar comparaciones**, algo que en Java no existe:

````python
nota = 8
print(7 <= nota <= 10)   # en Java: nota >= 7 && nota <= 10
````

## 7. Indentación y decisiones: `if / elif / else` (15 min)

En Python **la indentación no es opcional**: es la forma de indicar qué instrucciones pertenecen a un bloque. Reemplaza a las llaves `{ }` de Java.

| Java | Python |
|---|---|
| `if (nota >= 7) {` | `if nota >= 7:` |
| `} else if (nota >= 5) {` | `elif nota >= 5:` |
| `} else {` | `else:` |

- Se escriben **dos puntos `:`** al final de la condición.
- No se necesitan paréntesis alrededor de la condición.
- El bloque se indenta con **4 espacios**.

````python
nota = float(input("Ingrese la nota del estudiante: "))

if nota >= 7:
    print("Aprobado")
elif nota <= 5:
    print("Supletorio")
else:
    print("Reprobado")

print("Fin de la evaluación")   # sin indentación: está fuera del if
````

⚠️ Si la indentación es incorrecta, Python muestra un `IndentationError`. Ejecute la celda y observe el mensaje:

````python
saldo = 50
if saldo > 0:
    print("Saldo positivo")
````

### Operador ternario
En Java: `String estado = (saldo > 0) ? "Activo" : "Inactivo";`

````python
saldo = 50
estado = "Activo" if saldo > 0 else "Inactivo"
print(estado)
````


Ahora en Python:

````python
nombre = input("ingrese su nombre: ")
monto = float(input("Ingrese el monto a invertir: "))


if monto >100:
    meses = int(input("ingrese el numero de meses a invertir"))
    if meses < 6:
        interes= monto * 0.55 * meses/12
        print("El interes es:", interes)
    elif meses > 6 and meses<=11:
        interes= monto * 5.55 * meses/12 
        print("El interes es: ", interes)
    else:
        interes = monto*0.07 *(meses/12)
        print("el interes es:",interes)
else:
    print("Inversion rechazada")
````
