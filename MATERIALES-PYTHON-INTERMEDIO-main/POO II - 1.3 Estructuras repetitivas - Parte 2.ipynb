## 1. Repaso: números primos (10 min)

El reto de la clase anterior: mostrar los primos entre 2 y 50. Una solución usa un ciclo **dentro** de otro y el `else` del `for`.

````python
for n in range(2, 51):
    for d in range(2, int(n ** 0.5) + 1):
        if n % d == 0:
            break          # tiene un divisor: no es primo
    else:
        print(n, end=" ")  # el for interno terminó sin break: es primo
````

## 2. Recorrer diccionarios (20 min)

| Qué recorrer | Python | Java (`HashMap`) |
|---|---|---|
| Claves | `for k in d:` o `d.keys()` | `map.keySet()` |
| Valores | `d.values()` | `map.values()` |
| Clave y valor | `for k, v in d.items():` | `map.entrySet()` + `getKey()` / `getValue()` |

````python
tasas = {"ahorros": 0.02, "plazo fijo": 0.07, "infantil": 0.03}

for producto in tasas:
    print(producto)

print("Tasa promedio:", sum(tasas.values()) / len(tasas))

for producto, tasa in tasas.items():
    print(f"{producto:<12} {tasa:>6.1%}")
````

### Lista de diccionarios
Es la forma más común de representar **registros** (como las filas de una tabla) antes de usar clases.

````python
socios = [
    {"nombre": "Ana Pérez", "saldo": 1520.75, "activo": True},
    {"nombre": "Luis Mora", "saldo": 80.00, "activo": False},
    {"nombre": "María Torres", "saldo": 640.30, "activo": True},
]

print(f"{'Socio':<14}{'Saldo':>10}  Estado")
for s in socios:
    estado = "activo" if s["activo"] else "inactivo"
    print(f"{s['nombre']:<14}{s['saldo']:>10,.2f}  {estado}")
````

## 3. `zip`, `sorted` y `reversed` (15 min)

`zip` recorre **dos o más colecciones a la vez**, emparejando sus elementos. Se detiene cuando se acaba la más corta.

````python
nombres = ["Ana", "Luis", "María"]
saldos = [1520.75, 80.00, 640.30]

for nombre, saldo in zip(nombres, saldos):
    print(f"{nombre}: ${saldo:,.2f}")

# zip también sirve para crear un diccionario
cuentas = dict(zip(nombres, saldos))
print(cuentas)
````

````python
# sorted devuelve una lista nueva ordenada; reversed recorre al revés
for s in sorted(saldos, reverse=True):
    print(s, end="  ")
print()
for n in reversed(nombres):
    print(n, end="  ")
print()

# Ordenar los socios por saldo, de mayor a menor
for nombre, saldo in sorted(cuentas.items(), key=lambda par: par[1], reverse=True):
    print(f"{nombre:<6} {saldo:>9,.2f}")
````

> 💡 `key=lambda par: par[1]` indica **por qué ordenar**: por el segundo elemento (el saldo). Las funciones `lambda` se estudiarán en **1.4 Funciones**.

## 4. Ciclos anidados (20 min)

Un ciclo **dentro de otro**: por cada vuelta del ciclo externo, el interno da **todas** sus vueltas. Si el externo da 3 vueltas y el interno 4, el cuerpo se ejecuta 3 × 4 = 12 veces.

````python
for fila in range(1, 4):
    for col in range(1, 5):
        print(f"({fila},{col})", end=" ")
    print()   # salto de línea al terminar cada fila
````

### Matriz: depósitos por socio y por trimestre
Una lista de listas equivale a un arreglo bidimensional `double[][]` en Java.

````python
socios = ["Ana", "Luis", "María"]
depositos = [
    [300, 250, 400, 350],   # Ana
    [ 50,   0,  80,  60],   # Luis
    [200, 220, 180, 260],   # María
]

print(f"{'Socio':<6}" + "".join(f"{'T' + str(t):>6}" for t in range(1, 5)) + f"{'Total':>8}")
for nombre, fila in zip(socios, depositos):
    total = 0
    linea = f"{nombre:<6}"
    for monto in fila:
        linea += f"{monto:>6}"
        total += monto
    print(linea + f"{total:>8}")
````

````python
# Total por trimestre (por columnas)
for t in range(4):
    suma = 0
    for fila in depositos:
        suma += fila[t]
    print(f"Trimestre {t + 1}: {suma}")
````

### Romper un ciclo anidado
`break` solo sale del ciclo **más interno**. Para salir de los dos, se usa una variable bandera (o, más adelante, una función con `return`).

````python
objetivo = 0          # buscamos el primer trimestre sin depósito
encontrado = False
for i, fila in enumerate(depositos):
    for t, monto in enumerate(fila):
        if monto == objetivo:
            print(f"{socios[i]} no depositó en el trimestre {t + 1}")
            encontrado = True
            break
    if encontrado:
        break
````

## 5. Patrones con ciclos (25 min)

Casi todos los problemas con ciclos son una combinación de cinco patrones. Para cada uno, Python tiene además una **función integrada** que hace lo mismo en una línea.

| Patrón | Con ciclo | Función integrada |
|---|---|---|
| Contar | `c += 1` si cumple | `sum(1 for x in l if ...)`, `l.count(x)` |
| Acumular | `total += x` | `sum(l)` |
| Buscar | `if x == buscado: break` | `x in l`, `l.index(x)` |
| Máximo / mínimo | `if x > mayor: mayor = x` | `max(l)`, `min(l)` |
| Filtrar | `if ...: nueva.append(x)` | comprensión (sección 6) |

````python
saldos = [1520.75, 80.0, 640.3, 15.0, 2300.0, 410.0]

# Contar los saldos menores a $100
bajos = 0
for s in saldos:
    if s < 100:
        bajos += 1

# Máximo sin usar max()
mayor = saldos[0]
for s in saldos[1:]:
    if s > mayor:
        mayor = s

# Filtrar los saldos de $500 o más
altos = []
for s in saldos:
    if s >= 500:
        altos.append(s)

print("Saldos bajos:", bajos)
print("Mayor:", mayor, "| con max():", max(saldos))
print("Saldos altos:", altos)
````

### `any` y `all`
- `any(...)`: ¿**alguno** cumple?
- `all(...)`: ¿**todos** cumplen?

````python
print("¿Algún saldo negativo?", any(s < 0 for s in saldos))
print("¿Todos sobre $10?", all(s > 10 for s in saldos))
````

## 6. Comprensiones (30 min)

Una **comprensión** crea una colección nueva a partir de otra en **una sola línea**. Es una de las características más usadas de Python y **no tiene un equivalente directo en Java** (lo más parecido son los *streams*: `lista.stream().map(...).filter(...)`).

```
[expresión  for elemento in colección  if condición]
```

````python
saldos = [1520.75, 80.0, 640.3, 15.0, 2300.0, 410.0]

# Con ciclo
con_interes = []
for s in saldos:
    con_interes.append(round(s * 1.02, 2))

# Con comprensión: el mismo resultado
con_interes2 = [round(s * 1.02, 2) for s in saldos]

print(con_interes)
print(con_interes == con_interes2)
````

### Con condición (filtrar)

````python
altos = [s for s in saldos if s >= 500]
print(altos)

nombres = ["  ana pérez", "LUIS mora ", "maría torres"]
limpios = [n.strip().title() for n in nombres]
print(limpios)

# if-else dentro de la expresión (transformar, no filtrar)
etiquetas = ["alto" if s >= 500 else "bajo" for s in saldos]
print(etiquetas)
````

### Comprensión de diccionario y de conjunto

````python
socios = ["Ana", "Luis", "María"]
saldos = [1520.75, 80.00, 640.30]

# dict: {clave: valor for ...}
cuentas = {n: s for n, s in zip(socios, saldos)}
print(cuentas)

# Solo los socios con más de $100
con_fondos = {n: s for n, s in cuentas.items() if s > 100}
print(con_fondos)

# set: {valor for ...}  (sin repetidos)
cedulas = ["0102030405", "0912345678", "0103456789", "1104567890"]
provincias = {c[:2] for c in cedulas}
print(sorted(provincias))
````

### ¿Cuándo **no** usar una comprensión?
- Cuando la lógica tiene **varios pasos** o necesita `print`, `break` o `try`: use un ciclo normal.
- Cuando la línea queda tan larga que **cuesta leerla**.

> 📌 **Regla práctica:** si no se entiende en una lectura, escríbala como ciclo.

## 7. Caso práctico: tabla de amortización (25 min)

Un socio pide un crédito de **$1 200** a **6 meses** con una tasa del **12 % anual** (1 % mensual). La cooperativa usa el **sistema francés**: la cuota es fija y se calcula así:

$$cuota = \frac{P \cdot i}{1 - (1 + i)^{-n}}$$

donde $P$ es el monto, $i$ la tasa mensual y $n$ el número de cuotas. En cada mes:
- **Interés** = saldo pendiente × $i$
- **Capital** = cuota − interés
- **Nuevo saldo** = saldo − capital

````python
monto = 1200.0
tasa_anual = 0.12
meses = 6

i = tasa_anual / 12
cuota = monto * i / (1 - (1 + i) ** -meses)

print(f"Cuota mensual: ${cuota:.2f}\n")
print(f"{'Mes':>3} {'Cuota':>9} {'Interés':>9} {'Capital':>9} {'Saldo':>10}")
print("-" * 44)

saldo = monto
total_interes = 0.0
for mes in range(1, meses + 1):
    interes = saldo * i
    capital = cuota - interes
    saldo -= capital
    total_interes += interes
    print(f"{mes:>3} {cuota:>9.2f} {interes:>9.2f} {capital:>9.2f} {abs(saldo):>10.2f}")

print("-" * 44)
print(f"Total pagado: ${cuota * meses:,.2f} | Intereses: ${total_interes:,.2f}")
````

> 💡 Se usa `abs(saldo)` en la última fila porque, por el redondeo de los `float`, el saldo final puede quedar en `-0.00`. En un sistema real se usaría `Decimal` (tema 1.2).

## 8. Menú interactivo: cajero de la cooperativa (15 min)

Juntamos todo: `while True`, `if`/`elif`, validación y acumuladores. Este es el esqueleto de muchos programas de consola.

````python
saldo = 100.0
movimientos = []

while True:
    print("\n=== Cooperativa TEC ===")
    print("1. Consultar saldo")
    print("2. Depositar")
    print("3. Retirar")
    print("4. Ver movimientos")
    print("0. Salir")
    opcion = input("Opción: ")

    if opcion == "1":
        print(f"Saldo: ${saldo:,.2f}")
    elif opcion == "2":
        monto = float(input("Monto: "))
        if monto > 0:
            saldo += monto
            movimientos.append(("depósito", monto))
        else:
            print("Monto inválido")
    elif opcion == "3":
        monto = float(input("Monto: "))
        if 0 < monto <= saldo:
            saldo -= monto
            movimientos.append(("retiro", monto))
        else:
            print("Monto inválido o saldo insuficiente")
    elif opcion == "4":
        for n, (tipo, m) in enumerate(movimientos, start=1):
            print(f"{n}. {tipo:<9} ${m:>8.2f}")
    elif opcion == "0":
        print("Gracias por usar el cajero.")
        break
    else:
        print("Opción no válida")
````

> 🔜 Si el usuario escribe letras donde va un número, `float()` lanza `ValueError` y el programa se detiene. Lo resolveremos en **1.5 Manejo de excepciones** con `try`/`except`.
