## 2. Decidir antes de repetir: `if`, `elif`, `else` (15 min)

Casi todo ciclo necesita una **condición**, así que primero veremos cómo se escribe una decisión en Python.

| Java | Python |
|---|---|
| `if (saldo >= 100) { ... }` | `if saldo >= 100:` |
| `else if (...)` | `elif ...:` |
| Llaves `{ }` marcan el bloque | La **sangría (4 espacios)** marca el bloque |
| `&&`, `\|\|`, `!` | `and`, `or`, `not` |

```python
saldo = 250.0

if saldo >= 1000:
    categoria = "Oro"
elif saldo >= 100:
    categoria = "Plata"
else:
    categoria = "Básica"

print(f"Saldo ${saldo:.2f} → categoría {categoria}")
```

### Condiciones compuestas
Python permite **encadenar comparaciones** como en matemáticas: `18 <= edad < 65`. En Java habría que escribir `edad >= 18 && edad < 65`.

```python
edad = 30
activo = True
tiene_deuda = False

if 18 <= edad < 65 and activo and not tiene_deuda:
    print("Puede solicitar un crédito")
else:
    print("No cumple los requisitos")
```

### Expresión condicional (operador ternario)
En Java: `String estado = saldo > 0 ? "con fondos" : "sin fondos";`

```python
saldo = 0
estado = "con fondos" if saldo > 0 else "sin fondos"
print(estado)
```

> ⚠️ **Error típico de quien viene de Java:** olvidar los dos puntos `:` al final del `if`, o mezclar espacios y tabulaciones en la sangría. Python muestra `SyntaxError` o `IndentationError`.

## 3. El ciclo `while` (20 min)

`while` repite un bloque **mientras** la condición sea verdadera. Se usa cuando **no sabemos cuántas veces** vamos a repetir.

| Java | Python |
|---|---|
| `while (i <= 5) { ...; i++; }` | `while i <= 5:` … `i += 1` |
| `i++` | No existe: se escribe `i += 1` |
| `do { ... } while (...)` | No existe: se usa `while True` con `break` |

```python
# Contador: cuenta de 1 a 5
i = 1
while i <= 5:
    print("Vuelta", i)
    i += 1          # si se olvida esta línea, el ciclo nunca termina
print("Fin, i vale", i)
```

### Acumulador
Un **acumulador** guarda una suma (o un producto) que crece en cada vuelta. Se inicia en `0` para sumar y en `1` para multiplicar.

**Ejemplo:** un socio deposita $50 cada mes. ¿En cuántos meses llega a $400?

```python
saldo = 0.0
meses = 0
while saldo < 400:
    saldo += 50
    meses += 1
print(f"Llega a ${saldo:.2f} en {meses} meses")
```

### Validar datos con `while`
Un uso muy común: **pedir un dato hasta que sea válido**. Ejecute la celda y pruebe con un monto negativo.

```python
monto = float(input("Monto a depositar: "))
while monto <= 0:
    print("El monto debe ser mayor que cero.")
    monto = float(input("Monto a depositar: "))
print(f"Depósito registrado: ${monto:.2f}")
```

### `while True` + `break`: el «do-while» de Python
Cuando el ciclo debe ejecutarse **al menos una vez**, se escribe `while True` y se sale con `break`. Escriba `salir` para terminar.

```python
total = 0.0
while True:
    dato = input("Monto (o 'salir'): ")
    if dato.lower() == "salir":
        break
    total += float(dato)
print(f"Total depositado: ${total:.2f}")
```

> ⚠️ **Ciclo infinito:** si la condición nunca se vuelve falsa, el ciclo no termina. En Jupyter se detiene con el botón **■ (Interrumpir)** o con `I`, `I`.

## 4. El ciclo `for` con `range` (25 min)

`for` se usa cuando **sabemos cuántas veces** repetir o queremos recorrer una colección. `range` genera la secuencia de números.

| Java | Python | Valores |
|---|---|---|
| `for (int i = 0; i < 5; i++)` | `for i in range(5):` | 0, 1, 2, 3, 4 |
| `for (int i = 1; i <= 12; i++)` | `for i in range(1, 13):` | 1 … 12 |
| `for (int i = 0; i < 10; i += 2)` | `for i in range(0, 10, 2):` | 0, 2, 4, 6, 8 |
| `for (int i = 5; i > 0; i--)` | `for i in range(5, 0, -1):` | 5, 4, 3, 2, 1 |

> 📌 Igual que en el slicing, **el final no se incluye**: `range(1, 13)` llega hasta 12.

```python
for i in range(5):
    print(i, end=" ")
print()
for mes in range(1, 13):
    print(mes, end=" ")
print()
for i in range(0, 10, 2):
    print(i, end=" ")
print()
for i in range(5, 0, -1):
    print(i, end=" ")
print("¡Despegue!")
```

### Ejemplo: ahorro programado
Un socio ahorra $80 al mes durante 6 meses. Mostremos el saldo de cada mes.

```python
cuota = 80.0
saldo = 0.0
for mes in range(1, 7):
    saldo += cuota
    print(f"Mes {mes:>2}: ${saldo:>8.2f}")
```

### Ejemplo: interés compuesto
$1 000 a plazo fijo con una tasa del 7 % anual. ¿Cuánto hay al final de cada año durante 5 años?

```python
capital = 1000.0
tasa = 0.07
for anio in range(1, 6):
    capital *= 1 + tasa
    print(f"Año {anio}: ${capital:,.2f}")
```

## 5. Recorrer colecciones con `for` (15 min)

En Python, `for` recorre **directamente los elementos** de una lista, una tupla, una cadena, un conjunto o un diccionario. Equivale al *for-each* de Java: `for (double d : depositos)`.

```python
depositos = [120.0, 45.5, 300.0, 80.0]

total = 0.0
for d in depositos:
    total += d
    print(f"Depósito: ${d:.2f}")
print(f"Total: ${total:.2f}")
```

```python
# Recorrer una cadena letra por letra
cedula = "0102030405"
suma = 0
for digito in cedula:
    suma += int(digito)
print("Suma de los dígitos:", suma)
```

### `enumerate`: el elemento **y** su posición
En Java se usa el índice `i` para tener ambos. En Python, `enumerate` entrega los dos a la vez. Con `start=1` empieza a contar desde 1.

```python
socios = ["Ana", "Luis", "María"]

for i, nombre in enumerate(socios, start=1):
    print(f"{i}. {nombre}")
```

> 💡 **Estilo Python:** evite `for i in range(len(socios)): print(socios[i])`. Funciona, pero es «Java escrito en Python». Prefiera `for nombre in socios` o `enumerate`.

## 6. `break`, `continue` y `else` en los ciclos (10 min)

| Instrucción | Qué hace | ¿En Java? |
|---|---|---|
| `break` | Sale del ciclo inmediatamente | Sí |
| `continue` | Salta a la siguiente vuelta | Sí |
| `else` del ciclo | Se ejecuta si el ciclo terminó **sin** `break` | No existe |

```python
# continue: ignorar los montos inválidos
montos = [100, -20, 50, 0, 75]
total = 0
for m in montos:
    if m <= 0:
        print("Se ignora:", m)
        continue
    total += m
print("Total válido:", total)
```

```python
# break + else: buscar una cédula en la lista
cedulas = ["0102030405", "0912345678", "1104567890"]
buscada = "0912345678"

for posicion, c in enumerate(cedulas):
    if c == buscada:
        print("Encontrada en la posición", posicion)
        break
else:
    print("No está registrada")
```

Cambie `buscada` por `"0000000000"` y vuelva a ejecutar: ahora se ejecuta el `else`.

## 7. Caso práctico: meta de ahorro de un socio (10 min)

Un socio quiere juntar **$1 500** para un curso. Deposita **$120 cada mes** y la cooperativa paga un **0,5 % mensual** sobre el saldo. Muestre el saldo mes a mes hasta alcanzar la meta y cuántos meses tarda.

```python
meta = 1500.0
cuota = 120.0
interes_mensual = 0.005

saldo = 0.0
mes = 0
print(f"{'Mes':>3} | {'Interés':>8} | {'Saldo':>9}")
print("-" * 26)
while saldo < meta:
    mes += 1
    interes = saldo * interes_mensual
    saldo += interes + cuota
    print(f"{mes:>3} | {interes:>8.2f} | {saldo:>9.2f}")

print(f"\nMeta alcanzada en {mes} meses con ${saldo:,.2f}")
```

## Soluciones de los ejercicios

```python
# Ejercicio 1
edades = [17, 25, 64, 70, 15, 40]
menores = adultos = mayores = 0
for e in edades:
    if e < 18:
        grupo = "menor"
        menores += 1
    elif e < 65:
        grupo = "adulto"
        adultos += 1
    else:
        grupo = "adulto mayor"
        mayores += 1
    print(f"{e:>3} años → {grupo}")
print(f"Menores: {menores} | Adultos: {adultos} | Adultos mayores: {mayores}")
```

```python
# Ejercicio 2
pares = impares = 0
for n in range(1, 101):
    if n % 2 == 0:
        pares += n
    else:
        impares += n
print("Suma de pares:", pares)      # 2550
print("Suma de impares:", impares)  # 2500
```

```python
# Ejercicio 3
saldo = 500.0
retiros = 0
while True:
    monto = float(input("Monto a retirar (0 para terminar): "))
    if monto == 0:
        break
    if monto > saldo:
        print("Saldo insuficiente")
        continue
    saldo -= monto
    retiros += 1
print(f"Saldo final: ${saldo:.2f} | Retiros: {retiros}")
```

```python
# Ejercicio 4
CLAVE = "coop2026"
for intento in range(1, 4):
    clave = input(f"Intento {intento} de 3 - clave: ")
    if clave == CLAVE:
        print("Bienvenido")
        break
    print("Clave incorrecta")
else:
    print("Cuenta bloqueada")
```
